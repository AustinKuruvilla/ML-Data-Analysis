# ML Data Analysis & Modeling

This notebook demonstrates a complete machine learning workflow including data analysis, feature engineering, and model training.

## Overview
- **Task**: Binary classification on synthetic dataset
- **Approach**: Exploratory Data Analysis (EDA) + Random Forest Classification
- **Skills**: Data cleaning, visualization, feature analysis, model evaluation

## Step 1: Load and Explore Data

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
from src.analysis import build_demo_dataset, train_and_eval

# Load the dataset
df = build_demo_dataset()
print("Dataset loaded successfully!")
print(f"\nDataset shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"\nFirst few rows:")
print(df)
print(f"\nDataset info:")
print(df.info())
print(f"\nBasic statistics:")
print(df.describe())
```

## Step 2: Feature Analysis

```python
# Feature correlation and distribution analysis
print("Feature Correlation with Target:")
features = ['feature1', 'feature2']
for feat in features:
    corr = df[feat].corr(df['label'])
    print(f"  {feat}: {corr:.4f}")

print("\nFeature Statistics by Class:")
for label in [0, 1]:
    print(f"\nClass {label}:")
    print(df[df['label'] == label][features].describe())
```

## Step 3: Data Preparation and Preprocessing

```python
# Check for missing values
print("Missing values per column:")
print(df.isnull().sum())

print("\nData quality check:")
print(f"  Total rows: {len(df)}")
print(f"  Complete rows: {df.dropna().shape[0]}")
print(f"  Duplicate rows: {df.duplicated().sum()}")
print("\nData is clean and ready for modeling!")
```

## Step 4: Model Training and Evaluation

```python
# Train Random Forest model
print("=" * 60)
print("TRAINING RANDOM FOREST CLASSIFIER")
print("=" * 60)
train_and_eval(df)
print("\nModel evaluation complete!")
```
